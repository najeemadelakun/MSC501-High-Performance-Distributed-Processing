````python

# ============================================================
# MSC501: HIGH-PERFORMANCE DISTRIBUTED PROCESSING SYSTEM
# STEP 1: PROJECT ENVIRONMENT SETUP
# ============================================================

from pathlib import Path
import sys
import os
import platform
import subprocess

# Project configuration
PROJECT_NAME = "MSC501-High-Performance-Distributed-Processing"
PROJECT_ROOT = Path("/content") / PROJECT_NAME

# Create project directories
directories = [
    "app",
    "app/core",
    "app/monitoring",
    "app/recovery",
    "app/workload",
    "tests",
    "benchmark",
    "reports",
    "docs",
    "deployment"
]

for directory in directories:
    (PROJECT_ROOT / directory).mkdir(parents=True, exist_ok=True)

# Create Python package initialisation files
init_files = [
    "app/__init__.py",
    "app/core/__init__.py",
    "app/monitoring/__init__.py",
    "app/recovery/__init__.py",
    "app/workload/__init__.py",
    "tests/__init__.py",
    "benchmark/__init__.py"
]

for file in init_files:
    (PROJECT_ROOT / file).touch(exist_ok=True)

# Move into project directory
os.chdir(PROJECT_ROOT)

# Display environment information
print("=" * 65)
print("PROJECT ENVIRONMENT INITIALISED")
print("=" * 65)

print(f"Project name: {PROJECT_NAME}")
print(f"Project location: {PROJECT_ROOT}")
print(f"Python version: {sys.version.split()[0]}")
print(f"Operating system: {platform.system()}")

print("\nPROJECT DIRECTORY STRUCTURE")
print("-" * 65)

for path in sorted(PROJECT_ROOT.rglob("*")):
    if path.is_dir():
        print(f"[DIR]  {path.relative_to(PROJECT_ROOT)}")
    else:
        print(f"[FILE] {path.relative_to(PROJECT_ROOT)}")

print("\nStep 1.1 completed successfully.")
````

````python

# ============================================================
# STEP 1.2: INSTALL PROJECT DEPENDENCIES
# ============================================================

packages = [
    "pytest",
    "pandas",
    "numpy",
    "matplotlib",
    "psutil",
    "pydantic",
    "fastapi",
    "uvicorn",
    "httpx",
    "reportlab",
    "networkx"
]

subprocess.check_call([
    sys.executable,
    "-m",
    "pip",
    "install",
    "-q",
    *packages
])

print("PACKAGE INSTALLATION COMPLETED")
print("-" * 50)

for package in packages:
    print(f"[OK] {package}")

print("\nStep 1.2 completed successfully.")
````

````python

# ============================================================
# STEP 2: PROJECT CONFIGURATION AND DOCUMENTATION
# ============================================================

from pathlib import Path
import os

PROJECT_ROOT = Path("/content/MSC501-High-Performance-Distributed-Processing")
os.chdir(PROJECT_ROOT)

# ------------------------------------------------------------
# 1. CREATE REQUIREMENTS.TXT
# ------------------------------------------------------------

requirements = """pytest
pandas
numpy
matplotlib
psutil
pydantic
fastapi
uvicorn
httpx
reportlab
networkx
"""

(PROJECT_ROOT / "requirements.txt").write_text(
    requirements, encoding="utf-8"
)

# ------------------------------------------------------------
# 2. CREATE .GITIGNORE
# ------------------------------------------------------------

gitignore = """# Python cache
__pycache__/
*.py[cod]
*$py.class

# Virtual environments
venv/
.env/
.venv/

# Jupyter
.ipynb_checkpoints/

# Test and coverage
.pytest_cache/
.coverage
htmlcov/

# Temporary files
*.tmp
*.log
*.bak

# Generated databases
*.db
*.sqlite3

# Operating system
.DS_Store
Thumbs.db

# Environment secrets
.env
*.pem
*.key

# Git credentials and local configuration
credentials.json
"""

(PROJECT_ROOT / ".gitignore").write_text(
    gitignore, encoding="utf-8"
)

# ------------------------------------------------------------
# 3. CREATE README.MD
# ------------------------------------------------------------

readme = """# MSC501: High-Performance Distributed Processing

## Project Overview

A Python-based distributed processing system featuring concurrent
execution, fault tolerance, structured logging, performance
benchmarking, automated testing, and container deployment.

## Core Components

- Deterministic processing pipeline
- Input ingestion and validation
- Concurrent and thread-safe execution
- Structured JSON monitoring and logging
- Synthetic workload generation
- Fault injection and recovery
- Performance and latency benchmarking
- Automated testing
- Container deployment

## Project Structure

```text
app/
    core/
    monitoring/
    recovery/
    workload/
benchmark/
deployment/
docs/
reports/
tests/
requirements.txt
README.md
```

## Technology Stack

Python, pytest, NumPy, Pandas, Pydantic, FastAPI,
Matplotlib, ReportLab and Docker.

## Performance Evaluation

The system will be evaluated using throughput,
P50, P90, P99 and P99.9 latency measurements,
concurrency tests and controlled failure injection.

## Installation

Install dependencies using:

```bash
pip install -r requirements.txt
```

## Testing

Run the automated test suite using:

```bash
pytest -v
```

## Deployment

Docker deployment configuration will be provided
as part of the engineering deliverables.
"""

(PROJECT_ROOT / "README.md").write_text(
    readme, encoding="utf-8"
)

# ------------------------------------------------------------
# 4. VERIFY CREATED FILES
# ------------------------------------------------------------

print("=" * 65)
print("STEP 2: PROJECT CONFIGURATION COMPLETED")
print("=" * 65)

for filename in ["requirements.txt", ".gitignore", "README.md"]:
    file_path = PROJECT_ROOT / filename
    print(f"[OK] {filename} | {file_path.stat().st_size} bytes")

print("\nPROJECT ROOT:", PROJECT_ROOT)
print("\nStep 2.1 completed successfully.")
````
