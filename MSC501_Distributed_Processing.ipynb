````python

# ============================================================
# STEP 3.1: CORE INPUT VALIDATION MODULE
# File: app/core/validator.py
# ============================================================

from pathlib import Path

PROJECT_ROOT = Path("/content/MSC501-High-Performance-Distributed-Processing")

validator_code = '''
from pydantic import BaseModel, Field, ConfigDict, field_validator
from typing import Literal
from uuid import UUID


class ProcessingRequest(BaseModel):
    """Validated request submitted to the processing subsystem."""

    model_config = ConfigDict(
        extra="forbid",
        frozen=True
    )

    operation_id: UUID
    operation: Literal["ADD", "SUBTRACT", "MULTIPLY", "DIVIDE"]
    operand_a: float = Field(allow_inf_nan=False)
    operand_b: float = Field(allow_inf_nan=False)

    @field_validator("operand_a", "operand_b")
    @classmethod
    def validate_finite_numbers(cls, value):
        if not isinstance(value, (int, float)):
            raise ValueError("Operands must be numeric")

        return value

    def validate_operation(self):
        if self.operation == "DIVIDE" and self.operand_b == 0:
            raise ValueError("Division by zero is not permitted")

        return True
'''

file_path = PROJECT_ROOT / "app/core/validator.py"
file_path.write_text(validator_code.strip() + "\n", encoding="utf-8")

print("Created:", file_path)
print("Input validation module saved successfully.")
````

````python

# ============================================================
# STEP 3.2: DETERMINISTIC PROCESSING ENGINE
# File: app/core/processor.py
# ============================================================

processor_code = '''
from threading import RLock
from app.core.validator import ProcessingRequest


class ProcessingEngine:
    """Thread-safe deterministic processing engine."""

    def __init__(self):
        self._lock = RLock()
        self._results = {}

    def execute(self, request: ProcessingRequest):
        request.validate_operation()

        operation_id = str(request.operation_id)

        with self._lock:

            if operation_id in self._results:
                previous = self._results[operation_id]

                return {
                    "operation_id": operation_id,
                    "result": previous["result"],
                    "status": "DUPLICATE",
                    "committed": False
                }

            if request.operation == "ADD":
                result = request.operand_a + request.operand_b

            elif request.operation == "SUBTRACT":
                result = request.operand_a - request.operand_b

            elif request.operation == "MULTIPLY":
                result = request.operand_a * request.operand_b

            elif request.operation == "DIVIDE":
                result = request.operand_a / request.operand_b

            else:
                raise ValueError("Unsupported operation")

            record = {
                "operation_id": operation_id,
                "result": result
            }

            self._results[operation_id] = record

            return {
                "operation_id": operation_id,
                "result": result,
                "status": "SUCCESS",
                "committed": True
            }

    def get_committed_count(self):
        with self._lock:
            return len(self._results)

    def get_result(self, operation_id):
        with self._lock:
            return self._results.get(str(operation_id))
'''

file_path = PROJECT_ROOT / "app/core/processor.py"
file_path.write_text(processor_code.strip() + "\n", encoding="utf-8")

print("Created:", file_path)
print("Processing engine saved successfully.")
````

````python

# ============================================================
# STEP 4.1: CONCURRENT PROCESSING ARCHITECTURE
# ============================================================

from pathlib import Path

PROJECT_ROOT = Path("/content/MSC501-High-Performance-Distributed-Processing")

# Update the processing engine
processor_code = '''
from threading import RLock
from hashlib import sha256
import json

from app.core.validator import ProcessingRequest


class ProcessingEngine:
    """Thread-safe deterministic processing engine."""

    def __init__(self):
        self._lock = RLock()
        self._results = {}

    def _fingerprint(self, request):
        payload = {
            "operation": request.operation,
            "operand_a": request.operand_a,
            "operand_b": request.operand_b
        }

        encoded = json.dumps(
            payload,
            sort_keys=True,
            separators=(",", ":")
        ).encode("utf-8")

        return sha256(encoded).hexdigest()

    def _calculate(self, request):
        if request.operation == "ADD":
            return request.operand_a + request.operand_b
        elif request.operation == "SUBTRACT":
            return request.operand_a - request.operand_b
        elif request.operation == "MULTIPLY":
            return request.operand_a * request.operand_b
        elif request.operation == "DIVIDE":
            return request.operand_a / request.operand_b

        raise ValueError("Unsupported operation")

    def execute(self, request: ProcessingRequest):
        request.validate_operation()

        operation_id = str(request.operation_id)
        fingerprint = self._fingerprint(request)

        # Fast duplicate check
        with self._lock:
            previous = self._results.get(operation_id)

            if previous is not None:
                if previous["fingerprint"] != fingerprint:
                    raise ValueError(
                        "Operation ID conflict: different payload "
                        "submitted with an existing operation ID"
                    )

                return {
                    "operation_id": operation_id,
                    "result": previous["result"],
                    "status": "DUPLICATE",
                    "committed": False
                }

        # Execute outside the lock to permit parallel work
        result = self._calculate(request)

        # Recheck and commit atomically
        with self._lock:
            previous = self._results.get(operation_id)

            if previous is not None:
                if previous["fingerprint"] != fingerprint:
                    raise ValueError(
                        "Operation ID conflict: different payload "
                        "submitted with an existing operation ID"
                    )

                return {
                    "operation_id": operation_id,
                    "result": previous["result"],
                    "status": "DUPLICATE",
                    "committed": False
                }

            self._results[operation_id] = {
                "result": result,
                "fingerprint": fingerprint
            }

            return {
                "operation_id": operation_id,
                "result": result,
                "status": "SUCCESS",
                "committed": True
            }

    def get_committed_count(self):
        with self._lock:
            return len(self._results)

    def get_result(self, operation_id):
        with self._lock:
            return self._results.get(str(operation_id))
'''

(PROJECT_ROOT / "app/core/processor.py").write_text(
    processor_code.strip() + "\\n", encoding="utf-8"
)

# Create the concurrent execution module
concurrency_code = '''
from concurrent.futures import ThreadPoolExecutor, as_completed


class ConcurrentProcessor:
    """Manages concurrent execution of processing requests."""

    def __init__(self, engine, max_workers=8):
        if max_workers < 1:
            raise ValueError("max_workers must be at least 1")

        self.engine = engine
        self.max_workers = max_workers

    def process_batch(self, requests):
        """Process requests concurrently and preserve input order."""

        results = [None] * len(requests)

        with ThreadPoolExecutor(
            max_workers=self.max_workers
        ) as executor:

            future_map = {
                executor.submit(self.engine.execute, request): index
                for index, request in enumerate(requests)
            }

            for future in as_completed(future_map):
                index = future_map[future]
                results[index] = future.result()

        return results
'''

(PROJECT_ROOT / "app/core/concurrency.py").write_text(
    concurrency_code.strip() + "\\n", encoding="utf-8"
)

print("PASS: Updated thread-safe processing engine")
print("PASS: Created concurrent execution module")
print("PASS: Critical section reduced to transaction checks and commits")
````

````python

# ============================================================
# STEP 5.1: STRUCTURED JSON LOGGER
# File: app/monitoring/logger.py
# ============================================================

from pathlib import Path

PROJECT_ROOT = Path("/content/MSC501-High-Performance-Distributed-Processing")

logger_code = r'''
import json
import logging
import sys
from datetime import datetime, timezone


class JSONFormatter(logging.Formatter):
    """Format log records as structured JSON."""

    def format(self, record):
        entry = {
            "timestamp": datetime.now(timezone.utc).isoformat(),
            "level": record.levelname,
            "message": record.getMessage()
        }

        if hasattr(record, "event_data"):
            entry.update(record.event_data)

        return json.dumps(entry, default=str)


class StructuredLogger:
    """Thread-safe structured JSON logging interface."""

    def __init__(self, name="distributed_processing"):
        self.logger = logging.getLogger(name)
        self.logger.setLevel(logging.INFO)
        self.logger.propagate = False

        if not self.logger.handlers:
            handler = logging.StreamHandler(sys.stdout)
            handler.setFormatter(JSONFormatter())
            self.logger.addHandler(handler)

    def log_event(self, message, event_data, level="INFO"):
        log_level = getattr(logging, level.upper(), logging.INFO)

        self.logger.log(
            log_level,
            message,
            extra={"event_data": event_data}
        )
'''

path = PROJECT_ROOT / "app/monitoring/logger.py"
path.write_text(logger_code.strip() + "\n", encoding="utf-8")

print("PASS: Structured JSON logger created")
````

````python

# ============================================================
# STEP 5.2: PERFORMANCE METRICS MODULE
# File: app/monitoring/metrics.py
# ============================================================

metrics_code = r'''
from threading import Lock
from time import time
import numpy as np


class MetricsCollector:
    """Thread-safe collection of processing performance metrics."""

    def __init__(self):
        self._lock = Lock()
        self._latencies = []
        self._status_counts = {}
        self._start_time = time()

    def record(self, latency_us, status):
        with self._lock:
            self._latencies.append(latency_us)

            self._status_counts[status] = (
                self._status_counts.get(status, 0) + 1
            )

    def snapshot(self):
        with self._lock:
            latencies = list(self._latencies)
            status_counts = dict(self._status_counts)
            elapsed = max(time() - self._start_time, 1e-9)

        if latencies:
            percentiles = np.percentile(
                latencies, [50, 90, 99, 99.9]
            )
        else:
            percentiles = [0, 0, 0, 0]

        return {
            "total_requests": len(latencies),
            "throughput_ops_sec": len(latencies) / elapsed,
            "p50_latency_us": float(percentiles[0]),
            "p90_latency_us": float(percentiles[1]),
            "p99_latency_us": float(percentiles[2]),
            "p999_latency_us": float(percentiles[3]),
            "status_counts": status_counts
        }
'''

path = PROJECT_ROOT / "app/monitoring/metrics.py"
path.write_text(metrics_code.strip() + "\n", encoding="utf-8")

print("PASS: Performance metrics module created")
````

````python

# ============================================================
# STEP 5.3: MONITORED PROCESSING WRAPPER
# File: app/core/observed_processor.py
# ============================================================

observed_code = r'''
from time import perf_counter_ns
from uuid import uuid4

from app.monitoring.logger import StructuredLogger
from app.monitoring.metrics import MetricsCollector


class ObservedProcessor:
    """Adds structured logging and metrics to a processing engine."""

    def __init__(self, engine, logger=None, metrics=None):
        self.engine = engine
        self.logger = logger or StructuredLogger()
        self.metrics = metrics or MetricsCollector()

    def execute(self, request):
        trace_id = str(uuid4())
        start_ns = perf_counter_ns()

        try:
            result = self.engine.execute(request)

            status_code = (
                208 if result["status"] == "DUPLICATE" else 200
            )

            status = result["status"]
            level = "INFO"

            return result

        except ValueError as error:
            status_code = 400
            status = "FAILED"
            level = "WARNING"
            raise

        except Exception:
            status_code = 500
            status = "ERROR"
            level = "ERROR"
            raise

        finally:
            latency_us = (
                perf_counter_ns() - start_ns
            ) / 1000.0

            event = {
                "trace_id": trace_id,
                "operation_id": str(request.operation_id),
                "operation": request.operation,
                "latency_us": round(latency_us, 3),
                "status": status,
                "status_code": status_code
            }

            self.metrics.record(latency_us, status)

            self.logger.log_event(
                "Processing request completed",
                event,
                level
            )

    def get_committed_count(self):
        return self.engine.get_committed_count()

    def get_result(self, operation_id):
        return self.engine.get_result(operation_id)
'''

path = PROJECT_ROOT / "app/core/observed_processor.py"
path.write_text(observed_code.strip() + "\n", encoding="utf-8")

print("PASS: Monitored processing wrapper created")
````

````python

from pathlib import Path

PROJECT_ROOT = Path("/content/MSC501-High-Performance-Distributed-Processing")

RECOVERY_DIR = PROJECT_ROOT / "app" / "recovery"
RECOVERY_DIR.mkdir(parents=True, exist_ok=True)

recovery_code = '''
from app.recovery.fault_injector import FaultInjector


class RecoveryManager:
    """
    Delivers processing requests with bounded retries.

    Each request retains its original operation ID across retries,
    allowing the processing engine to enforce idempotency.
    """

    def __init__(self, engine, injector, max_attempts=5):

        if max_attempts < 1:
            raise ValueError("max_attempts must be at least 1")

        self.engine = engine
        self.injector = injector
        self.max_attempts = max_attempts

        self.successful_requests = 0
        self.recovered_requests = 0
        self.failed_requests = 0
        self.total_retries = 0

    def process_request(self, request):

        for attempt in range(1, self.max_attempts + 1):

            delivered = self.injector.deliver()

            if not delivered:

                if attempt < self.max_attempts:
                    self.total_retries += 1

                continue

            try:
                result = self.engine.execute(request)

                self.successful_requests += 1

                if attempt > 1:
                    self.recovered_requests += 1

                return {
                    "operation_id": str(request.operation_id),
                    "status": result["status"],
                    "attempts": attempt,
                    "recovered": attempt > 1,
                    "result": result
                }

            except Exception as error:

                self.failed_requests += 1

                return {
                    "operation_id": str(request.operation_id),
                    "status": "PROCESSING_ERROR",
                    "attempts": attempt,
                    "recovered": False,
                    "error": str(error)
                }

        self.failed_requests += 1

        return {
            "operation_id": str(request.operation_id),
            "status": "DELIVERY_FAILED",
            "attempts": self.max_attempts,
            "recovered": False
        }

    def process_batch(self, requests):

        results = [
            self.process_request(request)
            for request in requests
        ]

        return results

    def snapshot(self):

        return {
            "successful_requests": self.successful_requests,
            "recovered_requests": self.recovered_requests,
            "failed_requests": self.failed_requests,
            "total_retries": self.total_retries
        }
'''

recovery_file = RECOVERY_DIR / "recovery_manager.py"
recovery_file.write_text(recovery_code.strip() + "\n")

import ast

ast.parse(recovery_file.read_text())

print("Created:", recovery_file)
print("PASS: Recovery manager syntax verified")

import sys

if str(PROJECT_ROOT) not in sys.path:
    sys.path.insert(0, str(PROJECT_ROOT))

from app.core.processor import ProcessingEngine
from app.workload.generator import generate_workload
from app.recovery.fault_injector import FaultInjector
from app.recovery.recovery_manager import RecoveryManager

engine = ProcessingEngine()

injector = FaultInjector(
    failure_rate=0.30,
    seed=42
)

recovery = RecoveryManager(
    engine=engine,
    injector=injector,
    max_attempts=5
)

requests = generate_workload(1000, seed=2026)

print("\nFAULT RECOVERY TEST")
print("=" * 65)

results = recovery.process_batch(requests)

recovery_summary = recovery.snapshot()
fault_summary = injector.snapshot()

successful = sum(
    result["status"] in ["SUCCESS", "DUPLICATE"]
    for result in results
)

failed = sum(
    result["status"] not in ["SUCCESS", "DUPLICATE"]
    for result in results
)

committed = engine.get_committed_count()

print("Submitted requests:", len(requests))
print("Successful requests:", successful)
print("Recovered requests:", recovery_summary["recovered_requests"])
print("Failed requests:", failed)
print("Total delivery attempts:", fault_summary["total_attempts"])
print("Injected delivery failures:", fault_summary["injected_failures"])
print("Total retries:", recovery_summary["total_retries"])
print("Committed operations:", committed)

assert len(results) == 1000
assert successful + failed == 1000
assert committed == successful
assert recovery_summary["failed_requests"] == failed

print("\nPASS: All submitted requests accounted for")
print("PASS: No duplicate transaction commits")
print("PASS: Successful requests committed correctly")
print("PASS: Failed requests explicitly identified")
print("\nSTEP 7.2 RETRY AND RECOVERY VERIFIED")
````

````python
import sys
import time
import statistics
from pathlib import Path

from app.core.processor import ProcessingEngine
from app.core.concurrency import ConcurrentProcessor
from app.workload.generator import generate_workload

PROJECT_ROOT = Path(
    "/content/MSC501-High-Performance-Distributed-Processing"
)

print("=" * 70)
print("STEP 10.1: SEQUENTIAL AND CONCURRENT PERFORMANCE BENCHMARKING")
print("=" * 70)

# Benchmark configuration
WORKLOAD_SIZE = 5000
RANDOM_SEED = 42
REPETITIONS = 3
WORKER_COUNTS = [2, 4, 8]

# Generate reproducible workload
requests = generate_workload(
    size=WORKLOAD_SIZE,
    seed=RANDOM_SEED
)

assert len(requests) == WORKLOAD_SIZE

print("\nBENCHMARK CONFIGURATION")
print("-" * 70)
print(f"Workload size: {WORKLOAD_SIZE}")
print(f"Repetitions: {REPETITIONS}")
print(f"Worker configurations: {WORKER_COUNTS}")
print(f"Random seed: {RANDOM_SEED}")

benchmark_results = {}

# Sequential benchmark
print("\nRUNNING SEQUENTIAL BENCHMARK")
print("-" * 70)

sequential_times = []

for repetition in range(REPETITIONS):

    engine = ProcessingEngine()

    start_time = time.perf_counter()

    results = [
        engine.execute(request)
        for request in requests
    ]

    elapsed = time.perf_counter() - start_time

    assert len(results) == WORKLOAD_SIZE
    assert engine.get_committed_count() == WORKLOAD_SIZE

    sequential_times.append(elapsed)

    print(
        f"Run {repetition + 1}: "
        f"{elapsed:.6f} seconds"
    )

sequential_average = statistics.mean(sequential_times)

sequential_throughput = (
    WORKLOAD_SIZE / sequential_average
)

benchmark_results["Sequential"] = {
    "workers": 1,
    "average_time": sequential_average,
    "throughput": sequential_throughput,
    "speedup": 1.0
}

print("\nRUNNING CONCURRENT BENCHMARKS")
print("-" * 70)

for workers in WORKER_COUNTS:

    execution_times = []

    for repetition in range(REPETITIONS):

        engine = ProcessingEngine()

        processor = ConcurrentProcessor(
            engine,
            max_workers=workers
        )

        start_time = time.perf_counter()

        results = processor.process_batch(requests)

        elapsed = time.perf_counter() - start_time

        assert len(results) == WORKLOAD_SIZE
        assert engine.get_committed_count() == WORKLOAD_SIZE

        execution_times.append(elapsed)

        print(
            f"Workers: {workers}, "
            f"Run: {repetition + 1}, "
            f"Time: {elapsed:.6f} seconds"
        )

    average_time = statistics.mean(execution_times)

    throughput = WORKLOAD_SIZE / average_time

    speedup = sequential_average / average_time

    benchmark_results[f"{workers} Workers"] = {
        "workers": workers,
        "average_time": average_time,
        "throughput": throughput,
        "speedup": speedup
    }

# Display comparative results
print("\n" + "=" * 70)
print("PERFORMANCE BENCHMARK RESULTS")
print("=" * 70)

print(
    f"{'Configuration':<20}"
    f"{'Workers':<12}"
    f"{'Avg Time (s)':<18}"
    f"{'Throughput':<18}"
    f"{'Speedup':<12}"
)

print("-" * 80)

for name, result in benchmark_results.items():

    print(
        f"{name:<20}"
        f"{result['workers']:<12}"
        f"{result['average_time']:<18.6f}"
        f"{result['throughput']:<18.2f}"
        f"{result['speedup']:<12.3f}"
    )

# Final integrity verification
assert all(
    result["average_time"] > 0
    and result["throughput"] > 0
    for result in benchmark_results.values()
)

print("\nINTEGRITY VERIFICATION")
print("-" * 70)
print("PASS: Sequential benchmark completed")
print("PASS: Concurrent benchmarks completed")
print("PASS: All benchmark workloads committed")
print("PASS: Reproducible workload configuration")
print("PASS: Execution measurements collected")

print("\nSTEP 10.1 COMPLETED")
print("SEQUENTIAL AND CONCURRENT BENCHMARKING COMPLETED")
````

````python

import time
import random
import threading
import statistics
import psutil
import pandas as pd
import matplotlib.pyplot as plt
import numpy as np

from app.core.processor import ProcessingEngine
from app.core.concurrency import ConcurrentProcessor
from app.workload.generator import generate_workload

print("=" * 75)
print("STEP 10.4: CPU AND MEMORY UTILISATION ANALYSIS")
print("=" * 75)

# ------------------------------------------------------------
# EXPERIMENT CONFIGURATION
# ------------------------------------------------------------

WORKLOAD_SIZE = 50000
RANDOM_SEED = 42
REPETITIONS = 3
SAMPLE_INTERVAL = 0.05

CONFIGURATIONS = [
    ("Sequential", 1),
    ("2 Workers", 2),
    ("4 Workers", 4),
    ("8 Workers", 8)
]

requests = generate_workload(
    size=WORKLOAD_SIZE,
    seed=RANDOM_SEED
)

process = psutil.Process()

results = {
    name: []
    for name, workers in CONFIGURATIONS
}

# ------------------------------------------------------------
# RESOURCE MONITOR
# ------------------------------------------------------------

def monitor_resources(stop_event, memory_samples):

    while not stop_event.is_set():

        try:
            memory_mb = (
                process.memory_info().rss / (1024 ** 2)
            )

            memory_samples.append(memory_mb)

        except psutil.Error:
            break

        stop_event.wait(SAMPLE_INTERVAL)


# ------------------------------------------------------------
# WARM-UP
# ------------------------------------------------------------

print("\nWARM-UP PHASE")
print("-" * 75)

warmup_requests = generate_workload(
    size=1000,
    seed=100
)

for name, workers in CONFIGURATIONS:

    engine = ProcessingEngine()

    if workers == 1:
        output = [
            engine.execute(request)
            for request in warmup_requests
        ]
    else:
        processor = ConcurrentProcessor(
            engine,
            max_workers=workers
        )
        output = processor.process_batch(warmup_requests)

    assert len(output) == 1000
    assert engine.get_committed_count() == 1000

    print(f"PASS: {name} warm-up")


# ------------------------------------------------------------
# RESOURCE MEASUREMENT
# ------------------------------------------------------------

print("\nRESOURCE MEASUREMENT")
print("-" * 75)

for repetition in range(REPETITIONS):

    execution_order = CONFIGURATIONS.copy()

    random.Random(200 + repetition).shuffle(
        execution_order
    )

    print(f"\nREPETITION {repetition + 1}")

    for name, workers in execution_order:

        engine = ProcessingEngine()

        memory_samples = []
        stop_event = threading.Event()

        baseline_memory = (
            process.memory_info().rss / (1024 ** 2)
        )

        # Establish process CPU-time baseline
        cpu_before = process.cpu_times()

        monitor_thread = threading.Thread(
            target=monitor_resources,
            args=(stop_event, memory_samples),
            daemon=True
        )

        monitor_thread.start()

        start_time = time.perf_counter()

        try:

            if workers == 1:

                output = [
                    engine.execute(request)
                    for request in requests
                ]

            else:

                processor = ConcurrentProcessor(
                    engine,
                    max_workers=workers
                )

                output = processor.process_batch(requests)

        finally:

            elapsed = time.perf_counter() - start_time

            stop_event.set()
            monitor_thread.join()

        cpu_after = process.cpu_times()

        # ----------------------------------------------------
        # CALCULATE RESOURCE METRICS
        # ----------------------------------------------------

        cpu_time_used = (
            (cpu_after.user - cpu_before.user)
            +
            (cpu_after.system - cpu_before.system)
        )

        average_cpu_percent = (
            cpu_time_used / elapsed
        ) * 100

        final_memory = (
            process.memory_info().rss / (1024 ** 2)
        )

        peak_memory = max(
            memory_samples + [baseline_memory, final_memory]
        )

        memory_increase = max(
            0,
            peak_memory - baseline_memory
        )

        # ----------------------------------------------------
        # TRANSACTION INTEGRITY
        # ----------------------------------------------------

        assert len(output) == WORKLOAD_SIZE
        assert engine.get_committed_count() == WORKLOAD_SIZE

        results[name].append({

            "Execution Time (s)": elapsed,

            "CPU Utilisation (%)":
                average_cpu_percent,

            "CPU Time (s)":
                cpu_time_used,

            "Baseline Memory (MB)":
                baseline_memory,

            "Peak Memory (MB)":
                peak_memory,

            "Memory Increase (MB)":
                memory_increase,

            "Final Memory (MB)":
                final_memory

        })

        print(
            f"{name:<15} "
            f"Time: {elapsed:.3f}s | "
            f"CPU: {average_cpu_percent:.1f}% | "
            f"Peak RAM: {peak_memory:.2f} MB"
        )


# ------------------------------------------------------------
# STATISTICAL SUMMARY
# ------------------------------------------------------------

summary = []

for name, workers in CONFIGURATIONS:

    measurements = results[name]

    summary.append({

        "Configuration": name,

        "Workers": workers,

        "Median Execution Time (s)":
            statistics.median(
                x["Execution Time (s)"]
                for x in measurements
            ),

        "Average CPU Utilisation (%)":
            statistics.mean(
                x["CPU Utilisation (%)"]
                for x in measurements
            ),

        "Maximum CPU Utilisation (%)":
            max(
                x["CPU Utilisation (%)"]
                for x in measurements
            ),

        "Average Peak Memory (MB)":
            statistics.mean(
                x["Peak Memory (MB)"]
                for x in measurements
            ),

        "Maximum Peak Memory (MB)":
            max(
                x["Peak Memory (MB)"]
                for x in measurements
            ),

        "Average Memory Increase (MB)":
            statistics.mean(
                x["Memory Increase (MB)"]
                for x in measurements
            )

    })

summary_df = pd.DataFrame(summary)

print("\n" + "=" * 75)
print("CPU AND MEMORY UTILISATION SUMMARY")
print("=" * 75)

display(summary_df.round(3))


# ------------------------------------------------------------
# FIGURE 10.4A: CPU UTILISATION
# ------------------------------------------------------------

plt.figure(figsize=(10, 6))

bars = plt.bar(
    summary_df["Configuration"],
    summary_df["Average CPU Utilisation (%)"]
)

plt.title(
    "Figure 10.4A: Average CPU Utilisation by Processing Configuration"
)
plt.xlabel("Processing Configuration")
plt.ylabel("Average CPU Utilisation (%)")
plt.grid(axis="y", alpha=0.3)

for bar in bars:
    value = bar.get_height()

    plt.text(
        bar.get_x() + bar.get_width() / 2,
        value,
        f"{value:.1f}%",
        ha="center",
        va="bottom"
    )

plt.tight_layout()
plt.show()


# ------------------------------------------------------------
# FIGURE 10.4B: PEAK MEMORY CONSUMPTION
# ------------------------------------------------------------

plt.figure(figsize=(10, 6))

bars = plt.bar(
    summary_df["Configuration"],
    summary_df["Average Peak Memory (MB)"]
)

plt.title(
    "Figure 10.4B: Average Peak Memory Consumption"
)
plt.xlabel("Processing Configuration")
plt.ylabel("Peak Memory Usage (MB)")
plt.grid(axis="y", alpha=0.3)

for bar in bars:
    value = bar.get_height()

    plt.text(
        bar.get_x() + bar.get_width() / 2,
        value,
        f"{value:.1f} MB",
        ha="center",
        va="bottom"
    )

plt.tight_layout()
plt.show()


# ------------------------------------------------------------
# FIGURE 10.4C: MEMORY INCREASE
# ------------------------------------------------------------

plt.figure(figsize=(10, 6))

bars = plt.bar(
    summary_df["Configuration"],
    summary_df["Average Memory Increase (MB)"]
)

plt.title(
    "Figure 10.4C: Average Memory Increase During Processing"
)
plt.xlabel("Processing Configuration")
plt.ylabel("Memory Increase (MB)")
plt.grid(axis="y", alpha=0.3)

for bar in bars:
    value = bar.get_height()

    plt.text(
        bar.get_x() + bar.get_width() / 2,
        value,
        f"{value:.2f} MB",
        ha="center",
        va="bottom"
    )

plt.tight_layout()
plt.show()


# ------------------------------------------------------------
# FIGURE 10.4D: CPU UTILISATION AGAINST EXECUTION TIME
# ------------------------------------------------------------

plt.figure(figsize=(10, 6))

for _, row in summary_df.iterrows():

    plt.scatter(
        row["Average CPU Utilisation (%)"],
        row["Median Execution Time (s)"],
        s=100,
        label=row["Configuration"]
    )

    plt.annotate(
        row["Configuration"],
        (
            row["Average CPU Utilisation (%)"],
            row["Median Execution Time (s)"]
        ),
        xytext=(6, 6),
        textcoords="offset points"
    )

plt.title(
    "Figure 10.4D: CPU Utilisation versus Execution Time"
)
plt.xlabel("Average CPU Utilisation (%)")
plt.ylabel("Median Execution Time (seconds)")
plt.grid(True, alpha=0.3)
plt.legend()
plt.tight_layout()
plt.show()


# ------------------------------------------------------------
# FIGURE 10.4E: RESOURCE COMPARISON
# ------------------------------------------------------------

fig, axes = plt.subplots(1, 2, figsize=(13, 5))

axes[0].bar(
    summary_df["Configuration"],
    summary_df["Average CPU Utilisation (%)"]
)

axes[0].set_title("Average CPU Utilisation")
axes[0].set_xlabel("Configuration")
axes[0].set_ylabel("CPU Utilisation (%)")
axes[0].grid(axis="y", alpha=0.3)

axes[1].bar(
    summary_df["Configuration"],
    summary_df["Average Peak Memory (MB)"]
)

axes[1].set_title("Average Peak Memory")
axes[1].set_xlabel("Configuration")
axes[1].set_ylabel("Memory (MB)")
axes[1].grid(axis="y", alpha=0.3)

fig.suptitle(
    "Figure 10.4E: Comparative CPU and Memory Resource Consumption",
    fontsize=13
)

plt.tight_layout()
plt.show()


# ------------------------------------------------------------
# INTEGRITY VERIFICATION
# ------------------------------------------------------------

print("\n" + "=" * 75)
print("RESOURCE ANALYSIS VERIFICATION")
print("=" * 75)

assert len(summary_df) == 4

for name, workers in CONFIGURATIONS:
    assert len(results[name]) == REPETITIONS

    assert all(
        item["Execution Time (s)"] > 0
        for item in results[name]
    )

print("PASS: All four configurations measured")
print("PASS: Three repetitions completed per configuration")
print("PASS: CPU utilisation measurements collected")
print("PASS: Memory utilisation measurements collected")
print("PASS: All transactions committed successfully")
print("PASS: Five original figures generated")

print("\nSTEP 10.4 COMPLETED")
print("CPU AND MEMORY UTILISATION ANALYSIS COMPLETED")
````

````python

class LatencyRecordingEngine:

    def __init__(self, engine):
        self.engine = engine
        self.local = threading.local()
        self.thread_lists = []
        self.lock = threading.Lock()

    def execute(self, request):

        start = time.perf_counter_ns()

        try:
            return self.engine.execute(request)

        finally:

            latency_ms = (
                time.perf_counter_ns() - start
            ) / 1_000_000

            if not hasattr(self.local, "latencies"):

                self.local.latencies = []

                with self.lock:
                    self.thread_lists.append(
                        self.local.latencies
                    )

            self.local.latencies.append(latency_ms)

    def get_latencies(self):

        return [
            latency
            for thread_data in self.thread_lists
            for latency in thread_data
        ]

    def get_committed_count(self):
        return self.engine.get_committed_count()

    def get_result(self):
        return self.engine.get_result()
````

````python

import time
import random
import threading
import statistics
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

from app.core.processor import ProcessingEngine
from app.core.concurrency import ConcurrentProcessor
from app.workload.generator import generate_workload

print("=" * 75)
print("STEP 10.5: TRANSACTION LATENCY ANALYSIS")
print("=" * 75)

# ------------------------------------------------------------
# 1. EXPERIMENT CONFIGURATION
# ------------------------------------------------------------

WORKLOAD_SIZE = 50000
RANDOM_SEED = 42
REPETITIONS = 3

CONFIGURATIONS = [
    ("Sequential", 1),
    ("2 Workers", 2),
    ("4 Workers", 4),
    ("8 Workers", 8)
]

requests = generate_workload(
    size=WORKLOAD_SIZE,
    seed=RANDOM_SEED
)

# ------------------------------------------------------------
# 2. LOW-CONTENTION LATENCY RECORDING
# ------------------------------------------------------------

class LatencyRecordingEngine:

    def __init__(self, engine):

        self.engine = engine
        self.local = threading.local()
        self.thread_lists = []
        self.lock = threading.Lock()

    def execute(self, request):

        start = time.perf_counter_ns()

        try:
            return self.engine.execute(request)

        finally:

            latency_ms = (
                time.perf_counter_ns() - start
            ) / 1_000_000

            if not hasattr(self.local, "latencies"):

                self.local.latencies = []

                with self.lock:
                    self.thread_lists.append(
                        self.local.latencies
                    )

            self.local.latencies.append(latency_ms)

    def get_latencies(self):

        return [
            latency
            for thread_data in self.thread_lists
            for latency in thread_data
        ]

    def get_committed_count(self):
        return self.engine.get_committed_count()

    def get_result(self):
        return self.engine.get_result()


# ------------------------------------------------------------
# 3. WARM-UP PHASE
# ------------------------------------------------------------

print("\nWARM-UP PHASE")
print("-" * 75)

warmup_requests = generate_workload(
    size=1000,
    seed=100
)

for name, workers in CONFIGURATIONS:

    engine = LatencyRecordingEngine(
        ProcessingEngine()
    )

    if workers == 1:

        output = [
            engine.execute(request)
            for request in warmup_requests
        ]

    else:

        processor = ConcurrentProcessor(
            engine,
            max_workers=workers
        )

        output = processor.process_batch(
            warmup_requests
        )

    assert len(output) == 1000
    assert engine.get_committed_count() == 1000
    assert len(engine.get_latencies()) == 1000

    print(f"PASS: {name} warm-up")


# ------------------------------------------------------------
# 4. LATENCY MEASUREMENT
# ------------------------------------------------------------

print("\nMEASUREMENT PHASE")
print("-" * 75)

results = {
    name: []
    for name, workers in CONFIGURATIONS
}

for repetition in range(REPETITIONS):

    execution_order = CONFIGURATIONS.copy()

    random.Random(300 + repetition).shuffle(
        execution_order
    )

    print(f"\nREPETITION {repetition + 1}")

    for name, workers in execution_order:

        engine = LatencyRecordingEngine(
            ProcessingEngine()
        )

        start_time = time.perf_counter()

        if workers == 1:

            output = [
                engine.execute(request)
                for request in requests
            ]

        else:

            processor = ConcurrentProcessor(
                engine,
                max_workers=workers
            )

            output = processor.process_batch(
                requests
            )

        elapsed = time.perf_counter() - start_time

        latencies = np.array(
            engine.get_latencies(),
            dtype=float
        )

        # Transaction integrity
        assert len(output) == WORKLOAD_SIZE
        assert engine.get_committed_count() == WORKLOAD_SIZE
        assert len(latencies) == WORKLOAD_SIZE
        assert np.all(np.isfinite(latencies))
        assert np.all(latencies > 0)

        measurement = {

            "Execution Time (s)": elapsed,

            "Throughput (ops/sec)":
                WORKLOAD_SIZE / elapsed,

            "Mean Latency (ms)":
                np.mean(latencies),

            "P50 Latency (ms)":
                np.percentile(latencies, 50),

            "P90 Latency (ms)":
                np.percentile(latencies, 90),

            "P95 Latency (ms)":
                np.percentile(latencies, 95),

            "P99 Latency (ms)":
                np.percentile(latencies, 99),

            "Maximum Latency (ms)":
                np.max(latencies),

            "Raw Latencies (ms)":
                latencies

        }

        results[name].append(measurement)

        print(
            f"{name:<15} "
            f"Time: {elapsed:.3f}s | "
            f"Mean: {measurement['Mean Latency (ms)']:.6f}ms | "
            f"P99: {measurement['P99 Latency (ms)']:.6f}ms"
        )


# ------------------------------------------------------------
# 5. AGGREGATE RESULTS
# ------------------------------------------------------------

summary = []

for name, workers in CONFIGURATIONS:

    measurements = results[name]

    summary.append({

        "Configuration": name,

        "Workers": workers,

        "Median Execution Time (s)":
            statistics.median(
                x["Execution Time (s)"]
                for x in measurements
            ),

        "Median Throughput (ops/sec)":
            statistics.median(
                x["Throughput (ops/sec)"]
                for x in measurements
            ),

        "Mean Latency (ms)":
            statistics.median(
                x["Mean Latency (ms)"]
                for x in measurements
            ),

        "P50 Latency (ms)":
            statistics.median(
                x["P50 Latency (ms)"]
                for x in measurements
            ),

        "P90 Latency (ms)":
            statistics.median(
                x["P90 Latency (ms)"]
                for x in measurements
            ),

        "P95 Latency (ms)":
            statistics.median(
                x["P95 Latency (ms)"]
                for x in measurements
            ),

        "P99 Latency (ms)":
            statistics.median(
                x["P99 Latency (ms)"]
                for x in measurements
            ),

        "Maximum Latency (ms)":
            max(
                x["Maximum Latency (ms)"]
                for x in measurements
            )

    })

summary_df = pd.DataFrame(summary)

print("\n" + "=" * 75)
print("TRANSACTION LATENCY SUMMARY")
print("=" * 75)

display(summary_df.round(6))


# ------------------------------------------------------------
# 6. FIGURE 10.5A: MEAN LATENCY
# ------------------------------------------------------------

plt.figure(figsize=(10, 6))

bars = plt.bar(
    summary_df["Configuration"],
    summary_df["Mean Latency (ms)"]
)

plt.title(
    "Figure 10.5A: Mean Transaction Latency",
    fontsize=13
)
plt.xlabel("Processing Configuration")
plt.ylabel("Mean Latency (ms)")
plt.grid(axis="y", alpha=0.3)

for bar in bars:
    value = bar.get_height()

    plt.text(
        bar.get_x() + bar.get_width() / 2,
        value,
        f"{value:.5f}",
        ha="center",
        va="bottom"
    )

plt.tight_layout()
plt.show()


# ------------------------------------------------------------
# 7. FIGURE 10.5B: LATENCY PERCENTILES
# ------------------------------------------------------------

percentile_columns = [
    "P50 Latency (ms)",
    "P90 Latency (ms)",
    "P95 Latency (ms)",
    "P99 Latency (ms)"
]

plt.figure(figsize=(11, 6))

for column in percentile_columns:

    plt.plot(
        summary_df["Configuration"],
        summary_df[column],
        marker="o",
        linewidth=2,
        label=column
    )

plt.title(
    "Figure 10.5B: Transaction Latency Percentiles",
    fontsize=13
)
plt.xlabel("Processing Configuration")
plt.ylabel("Latency (ms)")
plt.grid(True, alpha=0.3)
plt.legend()
plt.tight_layout()
plt.show()


# ------------------------------------------------------------
# 8. FIGURE 10.5C: P99 LATENCY COMPARISON
# ------------------------------------------------------------

plt.figure(figsize=(10, 6))

bars = plt.bar(
    summary_df["Configuration"],
    summary_df["P99 Latency (ms)"]
)

plt.title(
    "Figure 10.5C: 99th Percentile Transaction Latency",
    fontsize=13
)
plt.xlabel("Processing Configuration")
plt.ylabel("P99 Latency (ms)")
plt.grid(axis="y", alpha=0.3)

for bar in bars:
    value = bar.get_height()

    plt.text(
        bar.get_x() + bar.get_width() / 2,
        value,
        f"{value:.5f}",
        ha="center",
        va="bottom"
    )

plt.tight_layout()
plt.show()


# ------------------------------------------------------------
# 9. FIGURE 10.5D: LATENCY DISTRIBUTION
# ------------------------------------------------------------

plt.figure(figsize=(11, 6))

plot_data = []

for name, workers in CONFIGURATIONS:

    combined = np.concatenate([
        item["Raw Latencies (ms)"]
        for item in results[name]
    ])

    plot_data.append(combined)

plt.boxplot(
    plot_data,
    tick_labels=[
        name for name, workers in CONFIGURATIONS
    ],
    showmeans=True,
    showfliers=False
)

plt.title(
    "Figure 10.5D: Transaction Latency Distribution",
    fontsize=13
)
plt.xlabel("Processing Configuration")
plt.ylabel("Transaction Latency (ms)")
plt.grid(axis="y", alpha=0.3)
plt.tight_layout()
plt.show()


# ------------------------------------------------------------
# 10. FIGURE 10.5E: THROUGHPUT VERSUS P99 LATENCY
# ------------------------------------------------------------

plt.figure(figsize=(10, 6))

for _, row in summary_df.iterrows():

    plt.scatter(
        row["Median Throughput (ops/sec)"],
        row["P99 Latency (ms)"],
        s=100,
        label=row["Configuration"]
    )

    plt.annotate(
        row["Configuration"],
        (
            row["Median Throughput (ops/sec)"],
            row["P99 Latency (ms)"]
        ),
        xytext=(6, 6),
        textcoords="offset points"
    )

plt.title(
    "Figure 10.5E: Throughput versus P99 Transaction Latency",
    fontsize=13
)
plt.xlabel("Median Throughput (operations/sec)")
plt.ylabel("P99 Latency (ms)")
plt.grid(True, alpha=0.3)
plt.legend()
plt.tight_layout()
plt.show()


# ------------------------------------------------------------
# 11. INTEGRITY VERIFICATION
# ------------------------------------------------------------

print("\n" + "=" * 75)
print("LATENCY ANALYSIS VERIFICATION")
print("=" * 75)

assert len(summary_df) == 4

for name, workers in CONFIGURATIONS:

    assert len(results[name]) == REPETITIONS

    for item in results[name]:

        assert len(item["Raw Latencies (ms)"]) == WORKLOAD_SIZE

        assert (
            item["P50 Latency (ms)"]
            <= item["P90 Latency (ms)"]
            <= item["P95 Latency (ms)"]
            <= item["P99 Latency (ms)"]
            <= item["Maximum Latency (ms)"]
        )

print("PASS: All four configurations evaluated")
print("PASS: Three repetitions completed per configuration")
print("PASS: 50,000 transaction latencies recorded per repetition")
print("PASS: Mean and percentile latencies calculated")
print("PASS: Latency measurements are finite and positive")
print("PASS: Transaction integrity verified")
print("PASS: Five original figures generated")

print("\nSTEP 10.5 COMPLETED")
print("TRANSACTION LATENCY ANALYSIS COMPLETED")
````

````python
import sys
import time
import statistics
import pandas as pd

from pathlib import Path

PROJECT_ROOT = Path(
    "/content/MSC501-High-Performance-Distributed-Processing"
)

if str(PROJECT_ROOT) not in sys.path:
    sys.path.insert(0, str(PROJECT_ROOT))

from app.core.processor import ProcessingEngine
from app.workload.generator import generate_workload
from app.recovery.fault_injector import FaultInjector
from app.recovery.recovery_manager import RecoveryManager


print("=" * 75)
print("STEP 10.6.2: FAULT TOLERANCE AND RECOVERY PERFORMANCE BENCHMARK")
print("=" * 75)

# ------------------------------------------------------------
# EXPERIMENT CONFIGURATION
# ------------------------------------------------------------

WORKLOAD_SIZE = 5000
REPETITIONS = 3
MAX_ATTEMPTS = 5

FAILURE_RATES = [
    0.0,
    0.10,
    0.30,
    0.50,
    0.80,
    1.0
]

BASE_SEED = 42

print("\nEXPERIMENT CONFIGURATION")
print("-" * 75)
print(f"Workload size:       {WORKLOAD_SIZE:,}")
print(f"Repetitions:         {REPETITIONS}")
print(f"Maximum attempts:    {MAX_ATTEMPTS}")
print(f"Failure rates:       {FAILURE_RATES}")
print("Execution mode:      Sequential recovery processing")


# ------------------------------------------------------------
# RECOVERY BENCHMARK
# ------------------------------------------------------------

benchmark_records = []

for failure_rate in FAILURE_RATES:

    print(f"\nTesting failure rate: {failure_rate:.0%}")

    repetition_records = []

    for repetition in range(1, REPETITIONS + 1):

        seed = BASE_SEED + repetition

        engine = ProcessingEngine()

        injector = FaultInjector(
            failure_rate=failure_rate,
            seed=seed
        )

        manager = RecoveryManager(
            engine=engine,
            injector=injector,
            max_attempts=MAX_ATTEMPTS
        )

        requests = generate_workload(
            WORKLOAD_SIZE,
            seed=BASE_SEED
        )

        start_time = time.perf_counter()

        results = manager.process_batch(requests)

        elapsed_time = time.perf_counter() - start_time

        injector_stats = injector.snapshot()
        recovery_stats = manager.snapshot()

        successful = sum(
            result["status"] == "SUCCESS"
            for result in results
        )

        failed = sum(
            result["status"] != "SUCCESS"
            for result in results
        )

        recovered = recovery_stats["recovered_requests"]

        direct_success = successful - recovered

        total_attempts = injector_stats["total_attempts"]

        committed = engine.get_committed_count()

        unique_request_ids = len({
            str(request.operation_id)
            for request in requests
        })

        successful_ids = [
            result["operation_id"]
            for result in results
            if result["status"] == "SUCCESS"
        ]

        unique_successful_ids = len(set(successful_ids))

        # ----------------------------------------------------
        # INTEGRITY VALIDATION
        # ----------------------------------------------------

        assert len(results) == WORKLOAD_SIZE
        assert successful + failed == WORKLOAD_SIZE

        assert committed == successful

        assert unique_request_ids == WORKLOAD_SIZE
        assert unique_successful_ids == successful

        assert (
            injector_stats["successful_deliveries"]
            + injector_stats["injected_failures"]
            == total_attempts
        )

        assert total_attempts <= WORKLOAD_SIZE * MAX_ATTEMPTS

        # ----------------------------------------------------
        # PERFORMANCE METRICS
        # ----------------------------------------------------

        success_rate = successful / WORKLOAD_SIZE * 100

        failure_rate_actual = failed / WORKLOAD_SIZE * 100

        recovery_coverage = recovered / WORKLOAD_SIZE * 100

        recovered_share = (
            recovered / successful * 100
            if successful else 0
        )

        average_attempts = total_attempts / WORKLOAD_SIZE

        throughput = WORKLOAD_SIZE / elapsed_time

        record = {
            "failure_rate": failure_rate,
            "repetition": repetition,
            "execution_time_s": elapsed_time,
            "throughput_ops_sec": throughput,
            "successful_requests": successful,
            "failed_requests": failed,
            "direct_successes": direct_success,
            "recovered_requests": recovered,
            "success_rate_pct": success_rate,
            "failure_rate_pct": failure_rate_actual,
            "recovery_coverage_pct": recovery_coverage,
            "recovered_share_pct": recovered_share,
            "total_attempts": total_attempts,
            "total_retries": recovery_stats["total_retries"],
            "average_attempts": average_attempts,
            "committed_transactions": committed
        }

        repetition_records.append(record)
        benchmark_records.append(record)

        print(
            f"  Repetition {repetition}: "
            f"{elapsed_time:.4f}s | "
            f"Success: {successful:,} | "
            f"Recovered: {recovered:,} | "
            f"Failed: {failed:,} | "
            f"Committed: {committed:,}"
        )

    print("  PASS: Repetition integrity checks completed.")


# ------------------------------------------------------------
# AGGREGATE RESULTS
# ------------------------------------------------------------

benchmark_df = pd.DataFrame(benchmark_records)

summary_records = []

for failure_rate in FAILURE_RATES:

    subset = benchmark_df[
        benchmark_df["failure_rate"] == failure_rate
    ]

    summary_records.append({
        "Failure Rate (%)": failure_rate * 100,

        "Median Time (s)": statistics.median(
            subset["execution_time_s"]
        ),

        "Median Throughput": statistics.median(
            subset["throughput_ops_sec"]
        ),

        "Success Rate (%)": statistics.median(
            subset["success_rate_pct"]
        ),

        "Recovery Coverage (%)": statistics.median(
            subset["recovery_coverage_pct"]
        ),

        "Recovered Share (%)": statistics.median(
            subset["recovered_share_pct"]
        ),

        "Average Attempts": statistics.mean(
            subset["average_attempts"]
        ),

        "Average Retries": statistics.mean(
            subset["total_retries"]
        ),

        "Average Failed Requests": statistics.mean(
            subset["failed_requests"]
        )
    })

summary_df = pd.DataFrame(summary_records)

print("\n" + "=" * 75)
print("RECOVERY PERFORMANCE SUMMARY")
print("=" * 75)

print(
    summary_df.round(3).to_string(index=False)
)


# ------------------------------------------------------------
# CONTROLLED REPLAY EXPERIMENT
# ------------------------------------------------------------

print("\n" + "=" * 75)
print("CONTROLLED REPLAY PERFORMANCE EVALUATION")
print("=" * 75)

replay_engine = ProcessingEngine()

replay_requests = generate_workload(
    WORKLOAD_SIZE,
    seed=BASE_SEED
)

# Stage 1: Force every initial delivery attempt to fail.
initial_injector = FaultInjector(
    failure_rate=1.0,
    seed=100
)

initial_manager = RecoveryManager(
    engine=replay_engine,
    injector=initial_injector,
    max_attempts=2
)

initial_start = time.perf_counter()

initial_results = initial_manager.process_batch(
    replay_requests
)

initial_time = time.perf_counter() - initial_start

initial_failed = [
    request
    for request, result in zip(
        replay_requests,
        initial_results
    )
    if result["status"] == "DELIVERY_FAILED"
]

initial_committed = replay_engine.get_committed_count()

assert len(initial_failed) == WORKLOAD_SIZE
assert initial_committed == 0

print(f"Initial requests:       {len(replay_requests):,}")
print(f"Initial failed:         {len(initial_failed):,}")
print(f"Initial committed:      {initial_committed:,}")
print(f"Initial execution time: {initial_time:.4f}s")


# Stage 2: Replay all failed requests without further delivery failures.
replay_injector = FaultInjector(
    failure_rate=0.0,
    seed=200
)

replay_manager = RecoveryManager(
    engine=replay_engine,
    injector=replay_injector,
    max_attempts=2
)

replay_start = time.perf_counter()

replay_results = replay_manager.process_batch(
    initial_failed
)

replay_time = time.perf_counter() - replay_start

replay_successes = sum(
    result["status"] == "SUCCESS"
    for result in replay_results
)

final_committed = replay_engine.get_committed_count()

original_ids = {
    str(request.operation_id)
    for request in replay_requests
}

replayed_ids = {
    result["operation_id"]
    for result in replay_results
}

assert replay_successes == WORKLOAD_SIZE
assert replayed_ids == original_ids
assert final_committed == WORKLOAD_SIZE
assert len(replayed_ids) == WORKLOAD_SIZE

print("\nREPLAY RESULTS")
print("-" * 75)
print(f"Requests replayed:      {len(initial_failed):,}")
print(f"Replay successes:       {replay_successes:,}")
print(f"Replay execution time:  {replay_time:.4f}s")
print(f"Replay throughput:      {replay_successes / replay_time:,.2f} ops/sec")
print(f"Final committed count:  {final_committed:,}")
print(f"Operation IDs preserved: {len(replayed_ids) == len(original_ids)}")


# ------------------------------------------------------------
# FINAL ACCEPTANCE CHECKS
# ------------------------------------------------------------

print("\n" + "=" * 75)
print("STEP 10.6.2: BENCHMARK ACCEPTANCE VERIFICATION")
print("=" * 75)

checks = {
    "All failure-rate configurations executed":
        len(benchmark_df) == len(FAILURE_RATES) * REPETITIONS,

    "All repetitions processed the expected workload":
        all(
            benchmark_df["successful_requests"]
            + benchmark_df["failed_requests"]
            == WORKLOAD_SIZE
        ),

    "Transaction accounting matches engine commits":
        all(
            benchmark_df["successful_requests"]
            == benchmark_df["committed_transactions"]
        ),

    "Delivery attempts remain within retry limits":
        all(
            benchmark_df["total_attempts"]
            <= WORKLOAD_SIZE * MAX_ATTEMPTS
        ),

    "Zero-failure configuration achieved full success":
        summary_df.loc[
            summary_df["Failure Rate (%)"] == 0,
            "Success Rate (%)"
        ].iloc[0] == 100.0,

    "Total-failure configuration achieved zero success":
        summary_df.loc[
            summary_df["Failure Rate (%)"] == 100,
            "Success Rate (%)"
        ].iloc[0] == 0.0,

    "Controlled replay recovered all failed requests":
        replay_successes == WORKLOAD_SIZE,

    "Controlled replay preserved operation IDs":
        replayed_ids == original_ids,

    "Final transaction count is correct":
        final_committed == WORKLOAD_SIZE
}

for description, passed in checks.items():
    print(f"{'PASS' if passed else 'FAIL'}: {description}")

assert all(checks.values()), "One or more acceptance checks failed."

print("\nSTEP 10.6.2 COMPLETED")
print("FAULT TOLERANCE AND REPLAY BENCHMARK VERIFIED")
````
